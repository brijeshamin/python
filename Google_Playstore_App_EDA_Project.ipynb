## 3. ***Data Wrangling***

### Data Wrangling Code

Data Wrangling : Data wrangling, also known as data cleaning or data preprocessing, is the process of transforming raw data into a clean and structured format that's suitable for analysis.

Data Wrangling Helps:

--> Improves Accuracy:
Clean data ensures that the insights or predictions you generate are reliable.

--> Enables Analysis:
You can't run graphs or models on text like "1,000+" — you need numbers.

--> Saves Time Later:
Clean data simplifies EDA, modeling, and visualization steps.

--> Prepares for Automation:
Clean datasets are essential for production-ready pipelines.

````python
# Step 1: Copy the original data before cleaning
apps_df_cleaned = apps_df.copy()
review_df_cleaned = review_df.copy()
````

````python
#Step 2: Clean the copied data

# Fill missing values in 'Rating' with the median
apps_df_cleaned['Rating'] = apps_df_cleaned['Rating'].fillna(apps_df_cleaned['Rating'].median())
apps_df_cleaned['Rating']
# .fillna : fills all NaN (missing) values with a specified value.
# .median(): Calculates the median (middle value) of all non-missing values in the 'Rating' column.
````

````python
# Drop rows with missing values in key columns
apps_df_cleaned.dropna(subset=['Type', 'Content Rating', 'Current Ver', 'Android Ver'], inplace=True)

#inplace= True : apply the change directly to the DataFrame
````

````python
# Drop rows with missing sentiment
review_df_cleaned.dropna(subset=['Sentiment'], inplace=True)
````

````python
# Convert 'Reviews' to numeric
apps_df_cleaned['Reviews'] = pd.to_numeric(apps_df_cleaned['Reviews'], errors='coerce')
apps_df_cleaned['Reviews']

#errors='coerce' : If any value cannot be converted, it will be replaced with NaN instead of throwing an error.
````

````python
# Clean 'Installs' column: remove '+' and ',' then convert to integer
apps_df_cleaned['Installs'] = apps_df_cleaned['Installs'].str.replace('[+,]', '', regex=True).astype(int)
````

````python
# Clean 'Price' column: remove '$' safely and convert to float
apps_df_cleaned['Price'] = apps_df_cleaned['Price'].astype(str).str.replace('$', '', regex=False)
apps_df_cleaned['Price'] = pd.to_numeric(apps_df_cleaned['Price'], errors='coerce')
````

````python
# Step 3: Preview the cleaned data
print("✅ First 5 rows of cleaned apps data:")
display(apps_df_cleaned.head())
````

````python
print("✅ First 5 rows of cleaned reviews data:")
display(review_df_cleaned.head())
````

The Installs column contains non-numeric characters like commas (,) and plus signs (+), e.g., "1,000,000+".

To use this column for analysis (like sorting or plotting), we must:

Remove unwanted characters

Removes the dollar sign from prices like '$4.99' → '4.99'

Converts strings into floating-point numbers: '4.99' → 4.99

Allows sorting, grouping, and plotting price data later

#### Chart - 19 - (SUBPLOT )
##### 1) Boxplot (Polarity vs Sentiment),
2) Histogram (Subjectivity),

3) Bar Chart (Avg Rating per Category),

4) Pie Chart (Content Rating Share)

````python
# Chart -19
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Create subplot grid (2x2)
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Polarity by Sentiment (Boxplot)",
        "Subjectivity Distribution (Histogram)",
        "Top 10 Categories by Avg Rating (Bar)",
        "Content Rating Share (Pie)"
    ),
    specs=[[{"type": "box"}, {"type": "xy"}], # type="box": Boxplot
           [{"type": "xy"}, {"type": "domain"}]]  # domain = pie chart
)
# 1. Boxplot – Sentiment Polarity by Sentiment
for sentiment in review_df_cleaned['Sentiment'].unique():
    fig.add_trace(
        go.Box(
            y=review_df_cleaned[review_df_cleaned['Sentiment'] == sentiment]['Sentiment_Polarity'],
            name=sentiment
        ),
        row=1, col=1
    )
# 2. Histogram – Subjectivity Distribution
fig.add_trace(
    go.Histogram(
        x=review_df_cleaned['Sentiment_Subjectivity'],
        nbinsx=30,
        marker_color='orange',
        name='Subjectivity'
    ),
    row=1, col=2
)
# 3. Bar Chart – Average Rating per Category (Top 10)
avg_rating = apps_df_cleaned.groupby('Category')['Rating'].mean().sort_values(ascending=False).head(10)
fig.add_trace(
    go.Bar(
        x=avg_rating.values,
        y=avg_rating.index,
        orientation='h',
        marker_color='skyblue',
        name='Avg Rating'
    ),
    row=2, col=1
)
# 4. Pie Chart – App Share by Content Rating
content_counts = apps_df_cleaned['Content Rating'].value_counts()
fig.add_trace(
    go.Pie(
        labels=content_counts.index,
        values=content_counts.values,
        textinfo='label+percent',
        hole=0.3,
        pull=[0.05]*len(content_counts),  # Slightly pulls each slice outward
        marker=dict(line=dict(color='#000000', width=1))  # Black border
    ),
    row=2, col=2
)
# Layout and Display
fig.update_layout(
    height=900,
    width=1100,
    title_text="Google Play Store – 4 Chart Dashboard (Plotly)",
    showlegend=False
)
fig.show()
````
